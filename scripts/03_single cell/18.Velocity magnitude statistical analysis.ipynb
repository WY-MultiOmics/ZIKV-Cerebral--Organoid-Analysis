```python
import scvelo as scv
import scanpy as sc
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib
import numpy as np

matplotlib.rcParams['pdf.fonttype'] = 42
matplotlib.rcParams['font.family'] = 'Arial'

adata.obs['velocity_length'] = np.linalg.norm(adata.layers['velocity'], axis=1)

sc.set_figure_params(dpi=300, facecolor='white')


fig, ax = plt.subplots(figsize=(6, 5))

sc.pl.violin(
    adata, 
    keys='velocity_length', 
    groupby='Group', 
    palette={'MOCK': '#3cc9c9', 'ZIKV': '#d4a5a5'},
    ax=ax,
    show=False,
    stripplot=True,
    color='darkgrey', 
    size=1.5,           
    alpha=0.3           
)


for artist in ax.get_children():
    
    if isinstance(artist, matplotlib.collections.PathCollection):
        artist.set_rasterized(True)
    if isinstance(artist, matplotlib.collections.PolyCollection):
        artist.set_rasterized(True)

ax.set_title('Velocity Magnitude: MOCK vs ZIKV', fontsize=14)
ax.set_ylabel('Velocity Length (L2 Norm)', fontsize=12)

save_path = '/mnt/d/WY_single_cell_analysis/results/scV_plt/Velocity_Magnitude_Hybrid.pdf'

plt.savefig(save_path, bbox_inches='tight', dpi=300, transparent=True)
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import matplotlib

matplotlib.rcParams['svg.fonttype'] = 'none'  
plt.rcParams['pdf.fonttype'] = 42             
plt.rcParams['font.family'] = 'Arial'         


df_plot = pd.DataFrame({
    'velocity_length': adata.obs['velocity_length'].values,
    'cell_type': adata.obs['cell_type'].values,
    'Group': adata.obs['Group'].values
})

df_plot['Group'] = df_plot['Group'].astype('category')
df_plot['cell_type'] = df_plot['cell_type'].astype('category')


plt.figure(figsize=(16, 8), dpi=300)

sns.violinplot(
    data=df_plot,
    x='cell_type',
    y='velocity_length',
    hue='Group',
    split=True,
    inner='quartile',
    palette={'MOCK': '#3cc9c9', 'ZIKV': '#d4a5a5'}
)


plt.xticks(rotation=90)
plt.title('Velocity Magnitude Comparison: MOCK vs ZIKV', fontsize=15)
plt.xlabel('Cell Type', fontsize=12)
plt.ylabel('Velocity Length (L2 Norm)', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.4)
plt.legend(title='Condition', loc='upper right')


save_path = '/mnt/d/WY_single_cell_analysis/results/scV_plt/Velocity_Final_Seaborn_Editable.svg'
plt.tight_layout()


plt.savefig(save_path, format='svg', bbox_inches='tight')
plt.savefig(save_path.replace('.svg', '.pdf'), format='pdf', bbox_inches='tight')

plt.show()
```
