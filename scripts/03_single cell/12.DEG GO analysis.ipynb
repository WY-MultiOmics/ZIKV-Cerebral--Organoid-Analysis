```python
import gseapy as gp
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import os

# 1. 路径设置
base_dir = "/mnt/d/WY_single_cell_analysis/results/final_annotation/annotation plt"
go_dir = os.path.join(base_dir, "ZIKV_group_DEG/GO_DEG")
os.makedirs(go_dir, exist_ok=True)

# 2. 修正后的绘图函数
def run_and_save_go_fixed(gene_list, title_text, file_prefix, color):
    if len(gene_list) < 5:
        print(f"⚠️ 基因数太少 ({len(gene_list)})，跳过: {title_text}")
        return

    # 运行 Enrichr - 注意 organism 参数已改为 'human'
    try:
        enr = gp.enrichr(gene_list=gene_list,
                         gene_sets=['GO_Biological_Process_2023'],
                         organism='human', 
                         outdir=None)
    except Exception as e:
        print(f"❌ 运行富集时出错: {e}")
        return

    res = enr.results
    res = res[res['Adjusted P-value'] < 0.05].head(10).copy()
    
    if res.empty:
        print(f"⚠️ {title_text} 未发现显著富集的条目")
        return

    # 绘图逻辑
    fig, ax = plt.subplots(figsize=(9, 5))
    res['-log10_p'] = -np.log10(res['Adjusted P-value'])
    ax.barh(res['Term'], res['-log10_p'], color=color, alpha=0.8, edgecolor='black', linewidth=0.5)
    
    ax.set_xlabel('-log10 Adjusted P-value', fontsize=12, fontweight='bold')
    ax.set_title(title_text, fontsize=14, fontweight='bold', pad=20)
    ax.invert_yaxis()
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    
    pdf_path = os.path.join(go_dir, f"{file_prefix}_Top10.pdf")
    plt.savefig(pdf_path, bbox_inches='tight', transparent=True)
    plt.show()
    plt.close()
    print(f"✅ 成功保存: {pdf_path}")

# 3. 执行分析
key_name = 'Intra_RG_Infect_vs_Active'

if key_name in adata.uns:
    df_deg = sc.get.rank_genes_groups_df(adata, group='ZIKV_Infected RG', key=key_name)
    up_genes = df_deg[(df_deg['logfoldchanges'] > 1) & (df_deg['pvals_adj'] < 0.05)]['names'].tolist()
    down_genes = df_deg[(df_deg['logfoldchanges'] < -1) & (df_deg['pvals_adj'] < 0.05)]['names'].tolist()

    run_and_save_go_fixed(up_genes, "GO: Up-regulated (Infected RG)", "GO_RG_Up", "#D62728")
    run_and_save_go_fixed(down_genes, "GO: Down-regulated (Infected RG)", "GO_RG_Down", "#1F77B4")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import gseapy as gp
import os

# 1. 路径与配置
base_dir = "/mnt/d/WY_single_cell_analysis/results/final_annotation/annotation plt/ZIKV_group_DEG/GO_DEG"
plt.rcParams['pdf.fonttype'] = 42
plt.rcParams['font.sans-serif'] = 'Arial'

# 定义分析清单：(Key, 组名, 简写)
all_intra_keys = [
    ('Intra_RG_Infect_vs_Active', 'ZIKV_Infected RG', 'RG'),
    ('Intra_IPCs_Infect_vs_Normal', 'ZIKV_Infected IPCs', 'IPCs'),
    ('Intra_Neurons_Infect_vs_Newborn', 'ZIKV_Infected Neurons', 'Neurons')
]

# 2. 创建大画布 (3行2列)
fig, axes = plt.subplots(3, 2, figsize=(16, 15))
plt.subplots_adjust(wspace=0.6, hspace=0.4)

def get_go_data(gene_list):
    """辅助函数：获取富集数据"""
    if len(gene_list) < 5: return None
    try:
        enr = gp.enrichr(gene_list=gene_list, gene_sets=['GO_Biological_Process_2023'],
                         organism='human', outdir=None)
        res = enr.results
        return res[res['Adjusted P-value'] < 0.05].head(10).copy()
    except:
        return None

# 3. 循环填充图像
row = 0
for key_name, group_name, label in all_intra_keys:
    if key_name in adata.uns:
        df_deg = sc.get.rank_genes_groups_df(adata, group=group_name, key=key_name)
        
        # 分离基因
        up_genes = df_deg[(df_deg['logfoldchanges'] > 1) & (df_deg['pvals_adj'] < 0.05)]['names'].tolist()
        down_genes = df_deg[(df_deg['logfoldchanges'] < -1) & (df_deg['pvals_adj'] < 0.05)]['names'].tolist()
        
        # 绘制 Up-regulated (左列)
        ax_up = axes[row, 0]
        res_up = get_go_data(up_genes)
        if res_up is not None and not res_up.empty:
            res_up['-log10_p'] = -np.log10(res_up['Adjusted P-value'])
            ax_up.barh(res_up['Term'], res_up['-log10_p'], color='#D62728', alpha=0.8, edgecolor='black')
            ax_up.set_title(f"Up: {label} (Infected vs Bystander)", fontsize=12, fontweight='bold')
            ax_up.invert_yaxis()
        
        # 绘制 Down-regulated (右列)
        ax_down = axes[row, 1]
        res_down = get_go_data(down_genes)
        if res_down is not None and not res_down.empty:
            res_down['-log10_p'] = -np.log10(res_down['Adjusted P-value'])
            ax_down.barh(res_down['Term'], res_down['-log10_p'], color='#1F77B4', alpha=0.8, edgecolor='black')
            ax_down.set_title(f"Down: {label} (Infected vs Bystander)", fontsize=12, fontweight='bold')
            ax_down.invert_yaxis()
        
        # 共有修饰
        for ax in [ax_up, ax_down]:
            ax.set_xlabel('-log10 Adj P-value')
            ax.spines['top'].set_visible(False)
            ax.spines['right'].set_visible(False)
            ax.tick_params(axis='y', labelsize=9)

    row += 1

# 4. 保存大图
combined_path = os.path.join(base_dir, "Combined_GO_Infection_Analysis.pdf")
plt.savefig(combined_path, bbox_inches='tight', transparent=True)
plt.show()

print(f"✅ 6合1 GO 组合图已保存至: {combined_path}")
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# 1. 精确匹配你打印出的组别名称
# 注意：这里将 'ZIKV_IPCs' 对应到我们想要观察的感染组
mapping = {
    'Mock_IPCs': 'Mock', 
    'ZIKV_IPCs': 'Infected',  # 修正：去掉了 _Infected 前缀
    'Mock_Newborn Neurons': 'Mock', 
    'ZIKV_Active-state Neurons': 'Bystander', 
    'ZIKV_Infected Neurons': 'Infected'
}

# 2. 选择要画的基因
genes_to_plot = ['TOP2A', 'CENPF', 'STMN2', 'MAP1B']

# 3. 绘图
fig, axes = plt.subplots(1, 4, figsize=(22, 6))
colors = ['#5A93C1', '#99B8C0', '#20458C'] # Mock(蓝), Bystander(浅蓝), Infected(深蓝)

for i, gene in enumerate(genes_to_plot):
    # 自动识别当前基因该画 IPCs 还是 Neurons 的对比
    relevant_keys = [k for k in mapping.keys() if ('IPCs' in k if i < 2 else 'Neurons' in k)]
    
    # 提取子集
    adata_tmp = adata[adata.obs['group_for_deg'].isin(relevant_keys)].copy()
    
    df = pd.DataFrame({
        'Expr': adata_tmp[:, gene].X.toarray().flatten() if hasattr(adata_tmp[:, gene].X, 'toarray') else adata_tmp[:, gene].X,
        'Status': adata_tmp.obs['group_for_deg'].map(mapping)
    })
    
    # 绘图 (Violin + Boxplot 组合)
    sns.violinplot(data=df, x='Status', y='Expr', ax=axes[i], 
                   order=['Mock', 'Bystander', 'Infected'] if i >= 2 else ['Mock', 'Infected'],
                   palette=colors, inner='quartile', cut=0)
    
    axes[i].set_title(f"$\it{{{gene}}}$", fontsize=16, fontweight='bold')
    axes[i].spines[['top', 'right']].set_visible(False)
    axes[i].set_ylabel("Normalized Expression")
    axes[i].set_xlabel("")

plt.suptitle("Neurodevelopmental Gene Expression: ZIKV vs Mock Control", fontsize=18, y=1.05)
plt.tight_layout()
plt.show()
```
