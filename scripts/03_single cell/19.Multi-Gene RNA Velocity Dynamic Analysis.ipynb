```python
import os
import numpy as np
import pandas as pd
import scanpy as sc
import seaborn as sns
import matplotlib.pyplot as plt
import matplotlib

# Step 3.5.1: Vector serialization environment tuning
matplotlib.rcParams['pdf.fonttype'] = 42
matplotlib.rcParams['svg.fonttype'] = 'none'
matplotlib.rcParams['font.family'] = 'Arial'
sns.set_style("white")

# Initialize computational paths and dynamic targets
file_path = "/mnt/d/.. adata_HNRNP_Clean_Integrated_Final.h5ad"
gene_output_dir = "/mnt/d/..HNRNP_Velo"
os.makedirs(gene_output_dir, exist_ok=True)

# Target gene matrix array list processing queue
target_genes = ['HNRNPD', 'HNRNPK', 'HNRNPU', 'HNRNPA2B1']

print("Loading data workspace matrix for automated phase-space calculations...")
adata = sc.read_h5ad(file_path)

# Loop processing sequence over target molecular targets
for gene in target_genes:
    if gene not in adata.var_names:
        print(f"[WARNING] Feature query '{gene}' not detected within active var matrix. Skipping target...")
        continue
        
    print(f"Extracting spliced/unspliced kinetics data states for feature: {gene}...")
    
    # Isolate independent analytical matrices by metadata condition rules
    mock_adata = adata[adata.obs['Group'] == 'MOCK']
    zikv_adata = adata[adata.obs['Group'] == 'ZIKV']
    
    # Structure computational data frames with flattened layer targets
    df_mock = pd.DataFrame({
        'Ms': mock_adata[:, gene].layers['Ms'].flatten(),
        'Mu': mock_adata[:, gene].layers['Mu'].flatten(),
        'Cell Type': mock_adata.obs['cell_type'].values,
        'Group': 'MOCK'
    })
    
    df_zikv = pd.DataFrame({
        'Ms': zikv_adata[:, gene].layers['Ms'].flatten(),
        'Mu': zikv_adata[:, gene].layers['Mu'].flatten(),
        'Cell Type': zikv_adata.obs['cell_type'].values,
        'Group': 'ZIKV'
    })
    
    # Unify matrices into an absolute target dataframe object
    df_all = pd.concat([df_mock, df_zikv], ignore_index=True)
    
    # Evaluate global dynamic boundary metrics with a 5% margin padding expansion
    max_ms = df_all['Ms'].max() * 1.05
    max_mu = df_all['Mu'].max() * 1.05
    
    # Step 3.5.2: Construct condition-stratified Facet Grid space
    g = sns.FacetGrid(
        df_all, 
        col="Group", 
        hue="Cell Type", 
        palette='viridis', 
        height=7, 
        aspect=1.1, 
        sharex=True, 
        sharey=True
    )
    
    # Stream high-density raw cell coordinates into rasterized paths to maintain high software rendering performance
    g.map(sns.scatterplot, "Ms", "Mu", s=40, alpha=0.7, edgecolor='none', rasterized=True)
    
    # Force coordinate alignment constraints to guarantee equal viewport scaling factors
    g.set(xlim=(0, max_ms), ylim=(0, max_mu)) 
    
    # Step 3.5.3: Professional annotation and aesthetics loop
    for ax in g.axes.flatten():
        ax.set_xlabel('Ms (Smoothed Spliced Expression)', fontsize=13, labelpad=8)
        ax.set_ylabel('Mu (Smoothed Unspliced Expression)', fontsize=13, labelpad=8)
        
        # Enforce highly precise grid references positioned behind raw points
        ax.grid(True, which='major', linestyle='-', alpha=0.3, zorder=-1)
        
        # Superimpose reference steady-state baseline vector
        ax.plot([0, max_ms], [0, max_ms], color='darkgray', linestyle='--', linewidth=1.5, zorder=0)
        sns.despine(ax=ax)
        
    g.add_legend(title="Cell Lineages Map")
    plt.subplots_adjust(top=0.88)
    g.fig.suptitle(f'Velocity Phase Portrait: {gene} (Normalized Dynamic Range)', fontsize=15, fontweight='bold')
    
    # Step 3.5.4: Serialization file execution output
    output_pdf_file = os.path.join(gene_output_dir, f'{gene}_1_Phase_Portrait_Normalized.pdf')
    plt.savefig(output_pdf_file, bbox_inches='tight', dpi=300)
    plt.close()
    
```
